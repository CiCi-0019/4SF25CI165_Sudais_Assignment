```python
import pandas as pd
import numpy as np

np.random.seed(42)

print("Pandas and NumPy imported successfully!")
```

```python
data = np.random.randint(1, 11, size=(5, 3))
df1 = pd.DataFrame(data, columns=["A", "B", "C"])
df1["A_times_B"] = df1["A"] * df1["B"]
print(df1)
```

## 2. Calculate row-wise and column-wise sums


```python
data = np.random.randint(1, 21, size=(4, 3))
df2 = pd.DataFrame(data, columns=["A", "B", "C"])

print("Original DataFrame:")
print(df2)

print("\nRow-wise sums:")
print(df2.sum(axis=1))

print("\nColumn-wise sums:")
print(df2.sum(axis=0))
```

## 3. Introduce NaN values and fill them with column means


```python
data = np.random.randint(1, 51, size=(5, 3)).astype(float)
df3 = pd.DataFrame(data, columns=["A", "B", "C"])

df3.iloc[1, 0] = np.nan
df3.iloc[3, 1] = np.nan
df3.iloc[4, 2] = np.nan

print("DataFrame with missing values:")
print(df3)

df3_filled = df3.fillna(df3.mean())
print("\nDataFrame after filling missing values with column means:")
print(df3_filled)
```

## 4. Drop rows containing any NaN values


```python
data = np.random.randint(1, 51, size=(6, 4)).astype(float)
df4 = pd.DataFrame(data, columns=["A", "B", "C", "D"])

df4.iloc[1, 2] = np.nan
df4.iloc[4, 0] = np.nan

print("Original DataFrame:")
print(df4)

df4_cleaned = df4.dropna()
print("\nDataFrame after dropping rows with missing values:")
print(df4_cleaned)
```

## 5. Data aggregation by Category


```python
categories = np.random.choice(["A", "B", "C"], size=12)
values = np.random.randint(10, 101, size=12)
df5 = pd.DataFrame({"Category": categories, "Value": values})

print("Original DataFrame:")
print(df5)

result5 = df5.groupby("Category")["Value"].agg(["sum", "mean"])
print("\nSum and mean of Value for each category:")
print(result5)
```

## 6. Calculate total sales for each product category


```python
products = np.random.choice(
    ["Laptop", "Phone", "Tablet", "Headphones", "Keyboard", "Mouse"],
    size=12
)
categories = np.random.choice(["Electronics", "Accessories", "Computers"], size=12)
sales = np.random.randint(500, 5001, size=12)

df6 = pd.DataFrame({"Product": products, "Category": categories, "Sales": sales})

print("Original Sales DataFrame:")
print(df6)

total_sales = df6.groupby("Category")["Sales"].sum()
print("\nTotal sales by category:")
print(total_sales)
```

## 7. Merge two DataFrames using a common column


```python
df7a = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Name": ["Aisha", "Rahul", "Sara", "Arjun"]
})

df7b = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Marks": [85, 92, 78, 88]
})

print("First DataFrame:")
print(df7a)

print("\nSecond DataFrame:")
print(df7b)

merged_df = pd.merge(df7a, df7b, on="Student_ID")
print("\nMerged DataFrame:")
print(merged_df)
```

## 8. MultiIndex DataFrame — indexing and slicing


```python
index = pd.MultiIndex.from_tuples(
    [
        ("Sales", "Aisha"),
        ("Sales", "Rahul"),
        ("IT", "Sara"),
        ("IT", "Arjun"),
        ("HR", "Meera"),
        ("HR", "Zaid")
    ],
    names=["Department", "Employee"]
)

df8 = pd.DataFrame(
    {"Salary": [30000, 32000, 45000, 47000, 28000, 29000]},
    index=index
)

print("MultiIndex DataFrame:")
print(df8)

print("\nAll employees in the IT department:")
print(df8.loc["IT"])

print("\nOne employee's record:")
print(df8.loc[("Sales", "Aisha")])
```

## 9. MultiIndex with Category and SubCategory


```python
df9 = pd.DataFrame({
    "Category": ["Food", "Food", "Food", "Clothing", "Clothing", "Clothing"],
    "SubCategory": ["Fruit", "Fruit", "Vegetable", "Men", "Women", "Men"],
    "Value": np.random.randint(10, 101, size=6)
})

df9_multi = df9.set_index(["Category", "SubCategory"])

print("MultiIndex DataFrame:")
print(df9_multi)

print("\nSum for each Category and SubCategory:")
print(df9_multi.groupby(level=["Category", "SubCategory"]).sum())
```

## 10. Apply a function to double every value


```python
data = np.random.randint(1, 21, size=(5, 3))
df10 = pd.DataFrame(data, columns=["A", "B", "C"])

def double_value(value):
    return value * 2

doubled_df = df10.map(double_value)

print("Original DataFrame:")
print(df10)

print("\nDataFrame after doubling every value:")
print(doubled_df)
```

## 11. Use a lambda function to calculate row totals


```python
data = np.random.randint(1, 51, size=(6, 3))
df11 = pd.DataFrame(data, columns=["A", "B", "C"])

df11["Total"] = df11.apply(
    lambda row: row["A"] + row["B"] + row["C"],
    axis=1
)

print(df11)
```

## 12. Convert text strings to uppercase


```python
texts = pd.Series([
    "barcelona",
    "pandas",
    "data analysis",
    "python programming",
    "machine learning"
])

print("Original text:")
print(texts)

uppercase_texts = texts.str.upper()
print("\nText in uppercase:")
print(uppercase_texts)
```
